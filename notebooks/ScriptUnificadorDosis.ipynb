# Script para unificar archivos de dosis de vacunación.

```python
import pandas as pd
import os
from pathlib import Path

base_path = Path(__file__).parent.parent if '__file__' in globals() else Path.cwd().parent
raw_data_path = f'{base_path}/data/raw/'
processed_data_path = f'{base_path}/data/processed/'
print(f"Ruta de datos en bruto: {raw_data_path}")
print(f"Ruta de datos procesados: {processed_data_path}")
input_directory = f'{raw_data_path}DGIS_dosis_srp'
output_filename = f'{processed_data_path}dosis_vacuna_2010-2023_unificado.csv'

# Lista para almacenar todos los dataframes
dataframes = []

# Leer todos los archivos CSV del 2010 al 2023
for year in range(2010, 2024):
    filename = f'dosis_vacuna_{year}.csv'
    filepath = Path(input_directory) / filename
    
    if filepath.exists():
        df = pd.read_csv(filepath)
        dataframes.append(df)
        print(f'✓ Cargado: {filename} - {len(df)} registros')
    else:
        print(f'✗ No encontrado: {filename}')
```

```python
# Unificar todos los dataframes
df_unificado = pd.concat(dataframes, ignore_index=True)

# Ordenar por año y mes
df_unificado = df_unificado.sort_values(['anio', 'mes']).reset_index(drop=True)

# Mostrar información del dataset unificado
print(f'\n--- RESUMEN ---')
print(f'Total de registros: {len(df_unificado)}')
print(f'Años incluidos: {df_unificado["anio"].min()} - {df_unificado["anio"].max()}')
print(f'Total de dosis (2010-2023): {df_unificado["total_dosis"].sum():,}')

# Guardar el dataset unificado
df_unificado.to_csv(output_filename, index=False)
print(f'\n✓ Archivo guardado: {output_filename}')

# Mostrar primeras y últimas filas
print('\nPrimeros registros:')
print(df_unificado.head())
print('\nÚltimos registros:')
print(df_unificado.tail())
```
